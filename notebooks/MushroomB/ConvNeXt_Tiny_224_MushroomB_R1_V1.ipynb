## **Section 2: Define model architecture**

```python
model_fn = partial(
    models.build_baseline_model,
    model_name=model_name,
    num_classes=NUM_CLASSES,
    device=device,
    freeze_backbone=False
)
model = model_fn()

criterion = nn.CrossEntropyLoss()
optimizer_fn = lambda m : optim.AdamW(m.parameters(), lr=train.train_config['base_lr'], weight_decay=train.train_config['weight_decay'])
scheduler_fn = partial(optim.lr_scheduler.CosineAnnealingLR, T_max=train.train_config['max_epochs'])

model_config = models.get_transform_config(model)
summary(model, (1,) + model_config['input_size'], device=device)
```

## **Section 3: Preparing the dataset**

**Configuration for training**

```python
img_size = model_config['img_size']
resize_size = int(img_size[0] / model_config['crop_pct'])
interp_mode = getattr(InterpolationMode, model_config['interpolation'].upper())
num_workers = min(4, os.cpu_count())

print(f'Image size: {img_size}')
print(f'Resize size: {resize_size}')
print(f'Interpolation mode: {interp_mode}')
print(f'Number of workers: {num_workers}')

reproducibility.global_seed(seed=train.train_config['seed'])
reproducibility.training_seed(seed=train.train_config['seed'])
g = reproducibility.get_generator(seed=train.train_config['seed'])
```

```python
train_transform = v2.Compose([
    v2.RandomResizedCrop(img_size, scale=(0.3, 1.0), interpolation=interp_mode),
    v2.RandomHorizontalFlip(p=0.5),
    v2.RandAugment(num_ops=2, magnitude=6, interpolation=interp_mode),
    v2.ToImage(),
    v2.ToDtype(torch.float32, scale=True),
    v2.Normalize(mean=model_config['mean'], std=model_config['std']),
    v2.RandomErasing(p=0.25, scale=(0.02, 0.33), ratio=(0.3, 3.3), value=0, inplace=False)
])

transform = v2.Compose([
    v2.Resize(resize_size, interpolation=interp_mode, max_size=None, antialias=True),
    v2.CenterCrop(img_size),
    v2.ToImage(),
    v2.ToDtype(torch.float32, scale=True),
    v2.Normalize(mean=model_config['mean'], std=model_config['std'])
])

dataloader_kwargs = {
    'batch_size': train.train_config['batch_size'],
    'num_workers': num_workers,
    'worker_init_fn': reproducibility.seed_worker,
    'generator': g,
    'pin_memory': True
}

datasets = datasets.Datasets(
    dataset_dir=dataset_dir,
    train_transform=train_transform,
    transform=transform,
    dataloader_kwargs=dataloader_kwargs,
)

datasets.print_summary()
test_loader = datasets.get_test_loader()
```

## **Section 4: Training the model**

```python
trainer = train.KFoldTrainer(
    n_splits=train.train_config['n_folds'],
    shuffle=True,
    random_state=train.train_config['seed'],
    model_name=model_name_path,
    dataset_name=dataset_name,
    base_dir=model_dir,
    first_metric=train.train_config['first_metric'],
    first_metric_mode=train.train_config['first_metric_mode'],
    secondary_metric=train.train_config['secondary_metric'],
    secondary_metric_mode=train.train_config['secondary_metric_mode'],
    device=device,
    criterion=criterion,
    n_classes=NUM_CLASSES,
    task=train.train_config['task'],
    average=train.train_config['average'],
    max_grad_norm=0.0
)
```

```python
fold_histories = trainer.fit_kfold(
    train_dataset=datasets.train_dataset,
    val_dataset=datasets.val_dataset,
    labels=datasets.all_classes,
    model_fn=model_fn,
    optimizer_fn=optimizer_fn,
    epochs=train.train_config['max_epochs'],
    dataloader_kwargs=dataloader_kwargs,
    callbacks_fn=callbacks_fn,
    scheduler_fn=scheduler_fn
)
```

##### **McNemar's test: single model vs ensemble, and single model vs baseline**

```python
from google.colab import drive
drive.mount('/content/drive')

import os
import json
import numpy as np
import pandas as pd

sc_drive_dir = '/content/drive/MyDrive/projectMushroom'
sc_dataset_name = 'MushroomB'
sc_n_folds, sc_weight_type, sc_alpha = 5, 'best', 0.05

def sc_get_model_dir(model_name_path, weights_root):
    return os.path.join(sc_drive_dir, f'{weights_root}/{sc_dataset_name}', f'{model_name_path}_{sc_dataset_name}')

# Model of this notebook
sc_model_name = 'ConvNeXt-Tiny' # เปลี่ยนเสมอ
sc_model_dir = sc_get_model_dir('convnext_tiny_224_r1_v1', 'modelWeights') # เปลี่ยนเสมอ

# Baseline (set to None in the ConvNeXt-T notebook)
sc_baseline_name = None
sc_baseline_dir = None

os.chdir(sc_drive_dir)
import src.ensembleVoting as ensembleVoting
import src.mcnemarTest as mcnemarTest
print(f'Model directory   : {sc_model_dir}')
print(f'Baseline directory: {sc_baseline_dir}')

def sc_load_single_and_ensemble(model_dir):
    """Correctness of the best single fold model and of the 5-fold soft-voting ensemble on the test set."""
    y_true, filepaths, _ = ensembleVoting.load_test_records(model_dir, weight_type=sc_weight_type)
    fold_probs = ensembleVoting.load_fold_probs(model_dir, sc_n_folds, sc_weight_type)
    if fold_probs.shape[1] != len(y_true):
        raise ValueError(f'[Error] {model_dir}: number of test samples differs')
    with open(os.path.join(model_dir, 'summary', 'best_fold_info.json')) as f:
        best_fold = int(json.load(f)['best_fold_num'])
    single_pred = fold_probs[best_fold - 1].argmax(axis=1)
    ens_pred = ensembleVoting.soft_voting(fold_probs).argmax(axis=1)
    return {'best_fold': best_fold, 'y_true': y_true, 'filepaths': filepaths,
            'Single': single_pred == y_true, 'Ensemble': ens_pred == y_true}

def sc_row(comparison, item, first, second, best_fold):
    res = mcnemarTest.mcnemar_test(second, first, alpha=sc_alpha)
    return {'Comparison': comparison, 'Model / Mode': item, 'Best fold': best_fold,
            'Acc 1 (%)': round(100 * first.mean(), 2), 'Acc 2 (%)': round(100 * second.mean(), 2),
            'Delta (pp)': round(100 * (first.mean() - second.mean()), 2),
            'Only 1': res['b'], 'Only 2': res['c'], 'Exact p-value': round(res['exact_p_value'], 4)}

sc_run = sc_load_single_and_ensemble(sc_model_dir)
sc_rows = [sc_row('Ensemble vs Single', sc_model_name, sc_run['Ensemble'], sc_run['Single'], sc_run['best_fold'])]

if sc_baseline_dir is not None:
    sc_base = sc_load_single_and_ensemble(sc_baseline_dir)
    if sc_run['filepaths'] != sc_base['filepaths'] or not np.array_equal(sc_run['y_true'], sc_base['y_true']):
        raise ValueError('[Error] Test images or labels of the two models are not aligned')
    for mode in ['Single', 'Ensemble']:
        sc_rows.append(sc_row(f'{sc_model_name} vs {sc_baseline_name}', mode, sc_run[mode], sc_base[mode],
                              f"{sc_run['best_fold']} / {sc_base['best_fold']}"))

sc_result_df = pd.DataFrame(sc_rows)
sc_save_dir = os.path.join(sc_model_dir, 'ensemble_addition', 'mcnemar')
os.makedirs(sc_save_dir, exist_ok=True)
sc_save_path = os.path.join(sc_save_dir, 'mcnemar_single_ensemble.csv')
sc_result_df.to_csv(sc_save_path, index=False)
print(f'[Logs] Saved to {sc_save_path}')
display(sc_result_df.style.hide(axis='index'))
```

##### **Edibility analysis on test set (Edible vs NotEdible)**

```python
from google.colab import drive
drive.mount('/content/drive')

import os
import zipfile
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import cv2
from sklearn.metrics import classification_report, confusion_matrix

ed_drive_dir = '/content/drive/MyDrive/projectMushroom'
ed_dataset_name = 'MushroomB'
ed_model_name = 'ConvNeXt-Tiny'
ed_model_dir = os.path.join(ed_drive_dir, f'modelWeights/{ed_dataset_name}', f'convnext_tiny_224_r1_v1_{ed_dataset_name}')
ed_model_ensem_dir = f'{ed_model_dir}/ensemble'
ed_save_dir = f'{ed_model_dir}/ensemble_addition/edibility'
ed_title = f'{ed_model_name} Ensemble on {ed_dataset_name} Dataset (Edibility)'
os.makedirs(ed_save_dir, exist_ok=True)

os.chdir(ed_drive_dir)
import src.visualization as visualization
import src.ensembleVoting as ensembleVoting
visualization.matplotlib_font(font_dir=f'{ed_drive_dir}/assets/fonts', prefer_name='Times-Roman')

ed_correct = pd.read_csv(f'{ed_model_ensem_dir}/ensem_correct_preds.csv')
ed_incorrect = pd.read_csv(f'{ed_model_ensem_dir}/ensem_incorrect_preds.csv')
ed_records = pd.concat([ed_correct, ed_incorrect], ignore_index=True)
print(f'[*] Correct: {len(ed_correct)} | Incorrect: {len(ed_incorrect)} | Total: {len(ed_records)}')
assert len(ed_records) == 600 and len(ed_incorrect) == 6, 'Unexpected number of test records'

ed_dataset_dir = f'/content/{ed_dataset_name}'
if not all(os.path.exists(p) for p in ed_incorrect['Filepaths']):
    with zipfile.ZipFile(f'{ed_drive_dir}/data/{ed_dataset_name}/{ed_dataset_name}_SPLIT.zip', 'r') as zip_ref:
        zip_ref.extractall(ed_dataset_dir, members=[m for m in zip_ref.namelist() if m.startswith('test/')])
print(f'[*] All misclassified images found: {all(os.path.exists(p) for p in ed_incorrect["Filepaths"])}')

# Classification report (Edibility)
ed_classes = ['Edible', 'NotEdible']           # index 0 = Edible, 1 = NotEdible

def ed_edibility(label):
    group = label.rsplit('-', 1)[-1]
    if group not in ed_classes:
        raise ValueError(f'[Error] Cannot read edibility from label: {label}')
    return ed_classes.index(group)

ed_y_true = ed_records['True_Label'].map(ed_edibility).to_numpy()
ed_y_pred = ed_records['Pred_Label'].map(ed_edibility).to_numpy()

ed_report = classification_report(ed_y_true, ed_y_pred, digits=4, target_names=ed_classes)

ed_report_dict = classification_report(ed_y_true, ed_y_pred, digits=4, target_names=ed_classes, output_dict=True)
ed_report_df = pd.DataFrame(ed_report_dict).transpose()
ed_report_df.to_csv(f'{ed_save_dir}/ensem_edibility_classification_report.csv')

print('Classification Report on Ensemble (Edibility)')
print('=' * 100)
print(ed_report)

ed_metrics = ensembleVoting.evaluate_labels(ed_y_true, ed_y_pred, task='multiclass', average='macro')
ed_metrics_df = pd.DataFrame([{k: v for k, v in ed_metrics.items() if k != 'loss'}])
ed_metrics_df.to_csv(f'{ed_save_dir}/ensem_edibility_test_perf.csv', index=False)
display(ed_metrics_df.style.hide(axis="index"))

# Confusion matrix (Edibility) without Normalize
ed_cm = confusion_matrix(ed_y_true, ed_y_pred, labels=[0, 1])
visualization.plot_confusion_matrix(
    cm=ed_cm,
    class_names=ed_classes,
    norm=False,
    cmap=plt.cm.YlGnBu,
    task='binary',
    figsize=(8, 6),
    fontsize=14,
    title=ed_title,
    save_path=ed_save_dir
)

# with Normalize
visualization.plot_confusion_matrix(
    cm=ed_cm,
    class_names=ed_classes,
    norm=True,
    cmap=plt.cm.YlGnBu,
    task='binary',
    figsize=(8, 6),
    fontsize=14,
    title=ed_title,
    save_path=ed_save_dir
)

def ed_plot_misclassified(df, img_size=(224, 224), img_base_size=3.5, titlesize=14, save_path=None):
    num_imgs = len(df)
    aspect_ratio = img_size[1] / img_size[0]
    fig, axes = plt.subplots(nrows=1, ncols=num_imgs,
                             figsize=((img_base_size * aspect_ratio) * num_imgs, img_base_size),
                             gridspec_kw={'wspace': 0.25, 'hspace': 0.02,
                                          'top': 0.95, 'bottom': 0.05,
                                          'left': 0.02, 'right': 0.98})
    axes = np.atleast_1d(axes)
    for i, (_, row) in enumerate(df.iterrows()):
        img = cv2.cvtColor(cv2.imread(row['Filepaths']), cv2.COLOR_BGR2RGB)
        ax = axes[i]
        ax.imshow(cv2.resize(img, img_size))
        ax.set_title(f"True: {row['True_Label']}\nPred: {row['Pred_Label']}", fontsize=titlesize, pad=10)
        ax.axis('off')
    if save_path:
        plt.savefig(f'{save_path}/misclassified_images.png', dpi=300, bbox_inches='tight')
        plt.savefig(f'{save_path}/misclassified_images.svg', bbox_inches='tight')
    plt.show()

ed_plot_misclassified(ed_incorrect, save_path=ed_save_dir)
```
